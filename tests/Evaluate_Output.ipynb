```python
import pandas as pd
```

```python
expected_clusters = pd.read_csv(
    "../Expected_results_for_csubellipsoidea/csubellipsoidea_Clust_v1_3.txt",
    delimiter="\t",
)
expected_genes = pd.read_csv(
    "../Expected_results_for_csubellipsoidea/csubellipsoidea_Gene_v1_3.txt",
    delimiter="\t",
)
resulting_clusters = pd.read_csv(
    "../Outputs/csubellipsoidea_Clust_v1_3.txt", delimiter="\t"
)
resulting_genes = pd.read_csv(
    "../Outputs/csubellipsoidea_Gene_v1_3.txt", delimiter="\t"
)

expected_clusters["Top n% cluster"] = expected_clusters["Top n% cluster"].astype("bool")
expected_clusters["Gene IDs"] = expected_clusters["Gene IDs"].apply(
    lambda x: x.split(", ")
)
resulting_clusters["Gene IDs"] = resulting_clusters["Gene IDs"].apply(
    lambda x: x.split(",")
)
for column in [
    "Enzyme",
    "Locally Duplicated",
    "Amines and Polyamines Metabolism",
    "Amino Acids Metabolism",
    "Carbohydrates Metabolism",
    "Cofactors Metabolism",
    "Detoxification Metabolism",
    "Energy Metabolism",
    "Fatty Acids and Lipids Metabolism",
    "Hormones Metabolism",
    "Inorganic Nutrients Metabolism",
    "Nitrogen-Containing Compounds",
    "Nucleotides Metabolism",
    "Phenylpropanoid Derivatives",
    "Polyketides",
    "Primary-Specialized Interface Metabolism",
    "Redox Metabolism",
    "Specialized Metabolism",
    "Sugar Derivatives",
    "Terpenoids",
]:
    expected_genes[column] = expected_genes[column].astype("bool")
expected_genes["EC"] = (
    expected_genes["EC"].fillna(value="").apply(lambda x: x.split(", "))
)
resulting_genes["EC-NUMBER"] = (
    resulting_genes["EC-NUMBER"].fillna(value="").apply(lambda x: x.split(","))
)
expected_genes["RXN"] = (
    expected_genes["RXN"].fillna(value="").apply(lambda x: x.split(", "))
)
resulting_genes["RXN"] = (
    resulting_genes["RXN"].fillna(value="").apply(lambda x: x.split(","))
)
```

```python
comparison_columns = [
    f"{x}_result"
    for x in resulting_clusters.columns
    if x in expected_clusters and x not in ["Chromosome", "Start bp"]
]
comparison_columns = [
    x
    for row in zip(
        comparison_columns,
        [
            f"{y}_expected"
            for y in expected_clusters.columns
            if y in resulting_clusters and y not in ["Chromosome", "Start bp"]
        ], strict=False,
    )
    for x in row
]
# comparison_columns
cluster_comparison = resulting_clusters.merge(
    expected_clusters, on=["Chromosome", "Start bp"], suffixes=("_result", "_expected")
)[comparison_columns]
```

```python
columns_to_compare = [
    col.split("_")[0] for col in cluster_comparison.columns if col.endswith("result")
]
```

```python
comparison_results = {
    column: cluster_comparison[
        cluster_comparison[f"{column}_result"]
        != cluster_comparison[f"{column}_expected"]
    ]
    for column in columns_to_compare
}
```

```python
[(col, len(val)) for col, val in comparison_results.items()]
```

```python
bp_diff = comparison_results["End bp"][
    [
        "ClusterID_result",
        "End bp_expected",
        "End bp_result",
        "number of Enzymes_result",
        "number of Enzymes_expected",
        "Number of genes_result",
        "Number of genes_expected",
        "Number of Hypothetical genes_result",
        "Number of Hypothetical genes_expected",
    ]
]
```

```python
bp_diff.loc[:, "diff_end"] = bp_diff["End bp_expected"] - bp_diff["End bp_result"]
bp_diff.loc[:, "diff_genes"] = (
    bp_diff["Number of genes_expected"] - bp_diff["Number of genes_result"]
)
bp_diff.loc[:, "diff_enzymes"] = (
    bp_diff["number of Enzymes_expected"] - bp_diff["number of Enzymes_result"]
)
bp_diff.loc[:, "diff_hypo"] = (
    bp_diff["Number of Hypothetical genes_expected"]
    - bp_diff["Number of Hypothetical genes_result"]
)
```

```python
gene_diff = resulting_genes.merge(
    expected_genes, on=resulting_genes.index, suffixes=("_result", "_expected")
)
gene_diff[gene_diff["Gene Name"] != gene_diff["Gene_Name"]]
```

```python
res_labels = [x for x in resulting_genes.columns if x not in expected_genes.columns]
exp_labels = [x for x in expected_genes.columns if x not in resulting_genes.columns][
    :-1
]
print(res_labels)
print(exp_labels)
res_labels += [
    f"{x}_result" for x in resulting_genes.columns if x in expected_genes.columns
]
exp_labels += [
    f"{x}_expected" for x in expected_genes.columns if x in resulting_genes.columns
]
```

```python
def compare_cols(df, col_1, col_2):
    return df[(df[col_1] != df[col_2]) & ((df[col_1].notna()) | (df[col_2].notna()))]


diff_gene_rows = {}

for res, exp in zip(res_labels, exp_labels, strict=False):
    diff_val = compare_cols(gene_diff, res, exp)
    if len(diff_val) > 0:
        diff_val = diff_val[[exp, res]]
        diff_gene_rows[exp] = diff_val
```

```python
# handle case where EC number list is ordered differently
diff_gene_rows["EC"] = diff_gene_rows["EC"][
    diff_gene_rows["EC"].apply(
        lambda x: all(y not in x["EC"] for y in x["EC-NUMBER"]), axis=1
    )
]
```

```python
# handle case where EC number list is ordered differently
diff_gene_rows["RXN_expected"] = diff_gene_rows["RXN_expected"][
    diff_gene_rows["RXN_expected"].apply(
        lambda x: all(y not in x["RXN_expected"] for y in x["RXN_result"]), axis=1
    )
]
```
